```python

# Environmental variables for further analysis.
environmental_features = [
    "Elevation",
    "Slope",
    "Aspect",
    "PlanCurv",
    "ProfCurv",
    "ExToppo",
    "LocalRelief",
    "DistChan",
    "DistRoad",
    "NearChanKsn",
    "NearChanSSP",
    "30Yr_AvPcP",
    "1Yr_TotPcP",
    "1Yr_PkMnthPcP",
    "Landuse_Co",
    "Geology",
    "Tectonic_Unit"
]

print(f"Number of environmental variables: {len(environmental_features)}")
print(environmental_features)
```

```python

# Select numerical environmental variables only.
numeric_features = [
    col for col in environmental_features
    if pd.api.types.is_numeric_dtype(df[col])
]

# Calculate correlations between numerical variables.
corr_matrix = df[numeric_features].corr().abs()

# Keep only the upper triangle to avoid duplicate pairs.
upper_triangle = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

# Convert the correlation matrix into a sorted table.
strong_correlations = (
    upper_triangle.stack()
    .reset_index()
)

strong_correlations.columns = [
    "Variable 1", "Variable 2", "Absolute Correlation"
]

strong_correlations = strong_correlations.sort_values(
    "Absolute Correlation", ascending=False
)

# Display the ten strongest relationships.
strong_correlations.head(10).round(3)
```

### Correlation Analysis — Findings

The strongest observed correlation was between annual total precipitation (`1Yr_TotPcP`) and peak monthly precipitation (`1Yr_PkMnthPcP`), with an absolute Pearson correlation of 0.924. Strong positive relationships were also observed between long-term average precipitation and annual precipitation (0.798), and between plan curvature and profile curvature (0.701).

These relationships suggest that some environmental variables may contain overlapping information. Further investigation is needed before deciding whether any predictors should be removed or combined. Because the analysis uses recorded landslide locations only, these correlations should not be interpreted as evidence of landslide causation or susceptibility.


## Handle Missing Values

A small number of environmental records contain missing values. To retain the recorded landslide locations, missing numerical values are imputed using their respective medians, while missing categorical values are labeled "Unknown". These replacements are applied only to the separate analysis dataset; the original dataset remains unchanged.

```python

# Create a new copy before handling missing values.
clean_environment_df = environment_df.copy()

# Identify numerical and categorical columns.
numeric_columns = clean_environment_df.select_dtypes(
    include="number"
).columns

categorical_columns = clean_environment_df.select_dtypes(
    include=["object", "str", "category"]
).columns

# Fill missing numerical values with each column's median.
for column in numeric_columns:
    clean_environment_df[column] = clean_environment_df[column].fillna(
        clean_environment_df[column].median()
    )

# Label missing categorical values as "Unknown".
for column in categorical_columns:
    clean_environment_df[column] = clean_environment_df[column].fillna(
        "Unknown"
    )

# Verify that no missing values remain.
print("Dataset shape:", clean_environment_df.shape)
print("Total missing values:", clean_environment_df.isnull().sum().sum())
```

```python
# Calculate correlations between numerical environmental variables
numeric_df = clean_environment_df.select_dtypes(include="number")

corr_matrix = numeric_df.corr()

# Keep only unique pairs and remove self-correlations
import numpy as np

upper_triangle = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

# List the 10 strongest correlations by absolute value
strong_correlations = (
    upper_triangle.stack()
    .rename("correlation")
    .reset_index()
    .rename(columns={"level_0": "Variable 1", "level_1": "Variable 2"})
)

strong_correlations["absolute_correlation"] = (
    strong_correlations["correlation"].abs()
)

strong_correlations = strong_correlations.sort_values(
    "absolute_correlation", ascending=False
).head(10)

display(strong_correlations.round(3))
```

### Key Findings: Correlations Between Environmental Variables

The correlation analysis identified several strong relationships among the numerical environmental variables associated with recorded landslide locations in Nepal.

The strongest positive correlation was between annual total precipitation and peak-month precipitation (\(r = 0.924\)). Long-term average precipitation was also positively correlated with annual total precipitation (\(r = 0.798\)) and peak-month precipitation (\(r = 0.736\)). These relationships suggest that the precipitation variables capture overlapping information.

Plan curvature and profile curvature showed a strong negative correlation (\(r = -0.701\)). Elevation was negatively correlated with long-term average precipitation (\(r = -0.619\)) and positively correlated with local relief (\(r = 0.606\)).

These findings reveal relationships among the environmental characteristics represented in the landslide inventory. However, they describe recorded landslide locations only. Without verified non-landslide comparison locations, the correlations cannot establish which characteristics distinguish landslide terrain from stable terrain or quantify landslide susceptibility. The meanings and units of the variables should also be checked against the original dataset documentation before drawing further conclusions.
