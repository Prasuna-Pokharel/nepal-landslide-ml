```python
import matplotlib.pyplot as plt  # Import Matplotlib for plotting
df["Year"].value_counts().sort_index().plot(
    kind="bar",
    figsize=(12, 5),
    title="Number of Landslides by Year"
)  # Plot the number of recorded landslides for each year

plt.xlabel("Year")  # Label the x-axis
plt.ylabel("Number of Landslides")  # Label the y-axis
plt.show()  # Display the chart
```
