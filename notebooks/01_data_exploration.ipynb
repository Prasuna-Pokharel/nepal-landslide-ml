
## Handle Missing Values

A small number of environmental records contain missing values. To retain the recorded landslide locations, missing numerical values are imputed using their respective medians, while missing categorical values are labeled "Unknown". These replacements are applied only to the separate analysis dataset; the original dataset remains unchanged.

```python

# Create a new copy before handling missing values.
clean_environment_df = environment_df.copy()

# Identify numerical and categorical columns.
numeric_columns = clean_environment_df.select_dtypes(
    include="number"
).columns

categorical_columns = clean_environment_df.select_dtypes(
    include=["object", "str", "category"]
).columns

# Fill missing numerical values with each column's median.
for column in numeric_columns:
    clean_environment_df[column] = clean_environment_df[column].fillna(
        clean_environment_df[column].median()
    )

# Label missing categorical values as "Unknown".
for column in categorical_columns:
    clean_environment_df[column] = clean_environment_df[column].fillna(
        "Unknown"
    )

# Verify that no missing values remain.
print("Dataset shape:", clean_environment_df.shape)
print("Total missing values:", clean_environment_df.isnull().sum().sum())
```

### Key Findings: Correlations Between Environmental Variables

The correlation analysis identified several strong relationships among the numerical environmental variables associated with recorded landslide locations in Nepal.

The strongest positive correlation was between annual total precipitation and peak-month precipitation (\(r = 0.924\)). Long-term average precipitation was also positively correlated with annual total precipitation (\(r = 0.798\)) and peak-month precipitation (\(r = 0.736\)). These relationships suggest that the precipitation variables capture overlapping information.

Plan curvature and profile curvature showed a strong negative correlation (\(r = -0.701\)). Elevation was negatively correlated with long-term average precipitation (\(r = -0.619\)) and positively correlated with local relief (\(r = 0.606\)).

These findings reveal relationships among the environmental characteristics represented in the landslide inventory. However, they describe recorded landslide locations only. Without verified non-landslide comparison locations, the correlations cannot establish which characteristics distinguish landslide terrain from stable terrain or quantify landslide susceptibility. The meanings and units of the variables should also be checked against the original dataset documentation before drawing further conclusions.

```python
import pandas as pd
import rasterio
from pyproj import Transformer

project = r"C:\Users\FM-PC-LT-333\Documents\nepal-landslide-ml"

inventory_path = (
    project + r"\data\raw\DataFile2 - ASMInventory_PointAttributes.txt"
)
dem_path = project + r"\data\processed\copernicus_dem_utm45n.tif"

# Load the landslide inventory
df = pd.read_csv(inventory_path, sep="\t", header=1, low_memory=False)
df.columns = df.columns.str.strip()

# Transform inventory coordinates to the DEM's CRS
transformer = Transformer.from_crs(
    "EPSG:32645", "EPSG:32645", always_xy=True
)

with rasterio.open(dem_path) as src:
    valid_points = 0
    outside_raster = 0
    masked_points = 0

    for x, y in zip(df["POINT_X"], df["POINT_Y"]):
        row, col = src.index(x, y)

        if not (0 <= row < src.height and 0 <= col < src.width):
            outside_raster += 1
            continue

        value = src.read(1, window=((row, row + 1), (col, col + 1)), masked=True)[0, 0]

        if value is not __import__("numpy").ma.masked and not __import__("numpy").ma.is_masked(value):
            valid_points += 1
        else:
            masked_points += 1

print("Total landslide points:", len(df))
print("Points on valid DEM pixels:", valid_points)
print("Points on masked pixels:", masked_points)
print("Points outside DEM:", outside_raster)
```

```python
import geopandas as gpd
import matplotlib.pyplot as plt
from shapely.geometry import box

# Load Nepal's national boundary
nepal = gpd.read_file(
    project / "data" / "processed" / "nepal_boundary.geojson"
)

# DEM bounds in EPSG:4326
dem_bounds = box(
    84.2995833333,  # West
    26.7404166667,  # South
    88.1895833333,  # East
    28.7204166667   # North
)

# Convert DEM bounds into a GeoDataFrame
dem_coverage = gpd.GeoDataFrame(
    {"area": ["DEM coverage"]},
    geometry=[dem_bounds],
    crs="EPSG:4326"
)

# Plot both boundaries
fig, ax = plt.subplots(figsize=(10, 8))

nepal.boundary.plot(
    ax=ax,
    color="black",
    linewidth=1.2,
    label="Nepal boundary"
)

dem_coverage.plot(
    ax=ax,
    color="orange",
    alpha=0.35,
    edgecolor="red",
    label="Current DEM coverage"
)

ax.set_title("Nepal Boundary and Current DEM Coverage")
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Convert the Nepal boundary and DEM extent to the inventory CRS
nepal_inventory_crs = nepal.to_crs(epsg=32645)
dem_inventory_crs = dem_coverage.to_crs(epsg=32645)

# Load the landslide inventory points
inventory = gpd.read_file(
    project / "data" / "raw" /
    "landslide_polygons" /
    "DataFile1 - 12838ASMInventory.shp"
)

# Make sure all layers use the same CRS
inventory = inventory.to_crs(epsg=32645)

# Count inventory points inside the DEM extent
dem_polygon = dem_inventory_crs.geometry.iloc[0]

inside_dem = inventory.geometry.within(dem_polygon)

print("Total inventory points:", len(inventory))
print("Points inside DEM extent:", inside_dem.sum())
print("Points outside DEM extent:", (~inside_dem).sum())
print(
    "Percentage inside:",
    round(inside_dem.mean() * 100, 2),
    "%"
)
```

```python
import rasterio
import geopandas as gpd
from rasterio.mask import mask

# Load the provisional study boundary
study_area = gpd.read_file(
    project / "data" / "processed" /
    "study_area_boundary.geojson"
)

# Open the projected DEM
dem_path = (
    project / "data" / "processed" /
    "copernicus_dem_utm45n.tif"
)

with rasterio.open(dem_path) as src:
    study_area_utm = study_area.to_crs(src.crs)

    clipped_dem, transform = mask(
        src,
        study_area_utm.geometry,
        crop=True,
        filled=False
    )

    dem_data = clipped_dem[0]

    total_pixels = dem_data.size
    valid_pixels = dem_data.count()
    nodata_pixels = total_pixels - valid_pixels

    print("Total pixels:", f"{total_pixels:,}")
    print("Valid DEM pixels:", f"{valid_pixels:,}")
    print("NoData pixels:", f"{nodata_pixels:,}")
    print(
        "Valid coverage:",
        f"{(valid_pixels / total_pixels) * 100:.2f}%"
    )
```

```python
import rasterio
import pandas as pd
import numpy as np
from pathlib import Path

# Load the existing training samples
training_df = pd.read_csv(
    "../data/processed/susceptibility_training_data.csv"
)

raster_files = {
    "DEM_Elevation": "copernicus_dem_utm45n.tif",
    "DEM_Slope": "slope_degrees.tif",
    "DEM_Aspect": "aspect_degrees.tif",
    "DEM_PlanCurv": "plan_curvature.tif",
    "DEM_ProfCurv": "profile_curvature.tif"
}

# Extract raster values at each sample coordinate
for column, filename in raster_files.items():

    raster_path = Path("../data/processed") / filename

    with rasterio.open(raster_path) as src:
        coordinates = list(zip(training_df["x"], training_df["y"]))
        values = [
            value[0] for value in src.sample(coordinates)
        ]

        training_df[column] = values

# Replace raster NoData values with NaN
feature_columns = list(raster_files.keys())

training_df[feature_columns] = training_df[feature_columns].replace(
    -9999, np.nan
)

# Check the extracted data
print("Dataset shape:", training_df.shape)
print("\nMissing values:")
print(training_df[feature_columns].isna().sum())

print("\nSample records:")
display(training_df.head())
```

## Preliminary Model Exploration

```python
from sklearn.model_selection import train_test_split

# Define predictors and target
X = training_df[features].copy()
y = training_df["label"].copy()

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("\nTraining class distribution:")
print(y_train.value_counts())
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier

models = {
    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(max_iter=1000, random_state=42))
    ]),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1,
        class_weight="balanced"
    ),

    "HistGradientBoosting": HistGradientBoostingClassifier(
        max_iter=200,
        learning_rate=0.1,
        max_leaf_nodes=31,
        random_state=42
    )
}

for name, model in models.items():
    model.fit(X_train, y_train)
    print(f"{name}: training complete")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

results = []

for name, model in models.items():

    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1 Score": f1_score(y_test, y_pred),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    })

results_df = pd.DataFrame(results).sort_values(
    by="ROC-AUC",
    ascending=False
)

display(results_df.round(4))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

best_model = models["HistGradientBoosting"]

y_pred_best = best_model.predict(X_test)

cm = confusion_matrix(y_test, y_pred_best)

print("Confusion Matrix:")
print(cm)

ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Background", "Landslide"]
).plot(cmap="Blues", values_format="d")

plt.title("HistGradientBoosting - Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
from sklearn.model_selection import GroupShuffleSplit

# Assign each sample to a 10 km spatial grid cell
block_size = 10000

spatial_groups = (
    (np.floor(training_df["x"] / block_size).astype(int).astype(str))
    + "_"
    + (np.floor(training_df["y"] / block_size).astype(int).astype(str))
)

# Split by spatial blocks, not individual samples
spatial_split = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    spatial_split.split(X, y, groups=spatial_groups)
)

X_train_spatial = X.iloc[train_idx]
X_test_spatial = X.iloc[test_idx]

y_train_spatial = y.iloc[train_idx]
y_test_spatial = y.iloc[test_idx]

print("Spatial training samples:", len(X_train_spatial))
print("Spatial testing samples:", len(X_test_spatial))

print("\nTraining class distribution:")
print(y_train_spatial.value_counts())

print("\nTesting class distribution:")
print(y_test_spatial.value_counts())

print("\nUnique spatial blocks:", spatial_groups.nunique())
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

spatial_results = []

spatial_models = {
    "Logistic Regression": models["Logistic Regression"],
    "Random Forest": models["Random Forest"],
    "HistGradientBoosting": models["HistGradientBoosting"]
}

for name, model in spatial_models.items():

    model.fit(X_train_spatial, y_train_spatial)

    y_pred = model.predict(X_test_spatial)
    y_prob = model.predict_proba(X_test_spatial)[:, 1]

    spatial_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test_spatial, y_pred),
        "Precision": precision_score(y_test_spatial, y_pred),
        "Recall": recall_score(y_test_spatial, y_pred),
        "F1 Score": f1_score(y_test_spatial, y_pred),
        "ROC-AUC": roc_auc_score(y_test_spatial, y_prob)
    })

spatial_results_df = pd.DataFrame(spatial_results).sort_values(
    by="ROC-AUC",
    ascending=False
)

display(spatial_results_df.round(4))
```

## Final Model Development and Evaluation

```python
features_circular = [
    "DEM_Elevation",
    "DEM_Slope",
    "Aspect_sin",
    "Aspect_cos",
    "DEM_PlanCurv",
    "DEM_ProfCurv"
]

X = training_df[features_circular]
y = training_df["label"]
```

```python
from sklearn.model_selection import GroupShuffleSplit

# Create 10 km spatial blocks.
block_size = 10000

spatial_groups = (
    (training_df["x"] // block_size).astype(str)
    + "_"
    + (training_df["y"] // block_size).astype(str)
)

# Split by spatial blocks.
gss = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    gss.split(X, y, groups=spatial_groups)
)

X_train_spatial = X.iloc[train_idx]
X_test_spatial = X.iloc[test_idx]

y_train_spatial = y.iloc[train_idx]
y_test_spatial = y.iloc[test_idx]

print("Training samples:", len(X_train_spatial))
print("Testing samples:", len(X_test_spatial))

print("\nTraining labels:")
print(y_train_spatial.value_counts())

print("\nTesting labels:")
print(y_test_spatial.value_counts())
```

```python
from sklearn.ensemble import (
    RandomForestClassifier,
    HistGradientBoostingClassifier
)
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# Define models.
models_circular = {
    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(max_iter=1000))
    ]),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1,
        class_weight="balanced"
    ),

    "HistGradientBoosting": HistGradientBoostingClassifier(
        max_iter=200,
        learning_rate=0.1,
        max_leaf_nodes=31,
        random_state=42
    )
}

# Train each model.
for name, model in models_circular.items():
    model.fit(X_train_spatial, y_train_spatial)
    print(f"{name} trained successfully.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

circular_results = []

for name, model in models_circular.items():

    y_pred = model.predict(X_test_spatial)
    y_prob = model.predict_proba(X_test_spatial)[:, 1]

    circular_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test_spatial, y_pred),
        "Precision": precision_score(y_test_spatial, y_pred),
        "Recall": recall_score(y_test_spatial, y_pred),
        "F1 Score": f1_score(y_test_spatial, y_pred),
        "ROC-AUC": roc_auc_score(y_test_spatial, y_prob)
    })

circular_results_df = pd.DataFrame(circular_results)

print(circular_results_df.round(4))
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

for ax, (name, model) in zip(axes, models_circular.items()):
    ConfusionMatrixDisplay.from_estimator(
        model,
        X_test_spatial,
        y_test_spatial,
        display_labels=["Background", "Landslide"],
        cmap="Blues",
        ax=ax
    )
    ax.set_title(name)

plt.tight_layout()
plt.show()
```

```python
best_model = models_circular["HistGradientBoosting"]

print("Selected model: HistGradientBoosting")
print("Features:", features_circular)
```

```python
from sklearn.inspection import permutation_importance

perm_importance = permutation_importance(
    best_model,
    X_test_spatial,
    y_test_spatial,
    n_repeats=10,
    random_state=42,
    scoring="roc_auc",
    n_jobs=-1
)

importance_df = pd.DataFrame({
    "Feature": features_circular,
    "Importance": perm_importance.importances_mean,
    "Std": perm_importance.importances_std
}).sort_values(
    by="Importance",
    ascending=False
)

print(importance_df.round(4))
```

```python
import numpy as np
from scipy.spatial import cKDTree

# Separate background and landslide coordinates
background_coords = background_df[["x", "y"]].to_numpy()

landslide_coords = training_df.loc[
    training_df["sample_type"] == "landslide_inventory",
    ["x", "y"]
].to_numpy()

# Find the nearest landslide point for each background point
tree = cKDTree(landslide_coords)
distances, _ = tree.query(background_coords, k=1)

print("Minimum distance:", round(distances.min(), 2), "m")
print("Median distance:", round(np.median(distances), 2), "m")
print("Mean distance:", round(distances.mean(), 2), "m")
print("Background points within 500 m:", np.sum(distances < 500))
print("Background points within 1 km:", np.sum(distances < 1000))
print("Background points within 5 km:", np.sum(distances < 5000))
```

## Generate the Final Susceptibility Map

```python
import rasterio
import numpy as np
import pandas as pd
import joblib

from pathlib import Path

# Load the trained model
model = joblib.load(
    "../models/hist_gradient_boosting_model.pkl"
)

# Define input rasters
raster_paths = {
    "DEM_Elevation": "../data/processed/copernicus_dem_utm45n.tif",
    "DEM_Slope": "../data/processed/slope_degrees.tif",
    "Aspect": "../data/processed/aspect_degrees.tif",
    "DEM_PlanCurv": "../data/processed/plan_curvature.tif",
    "DEM_ProfCurv": "../data/processed/profile_curvature.tif",
}

# Read raster data
arrays = {}
profile = None
valid_mask = None

for name, path in raster_paths.items():
    with rasterio.open(path) as src:
        data = src.read(1).astype(np.float32)

        if profile is None:
            profile = src.profile.copy()

        valid = (
            src.read_masks(1).astype(bool)
            & np.isfinite(data)
        )

        arrays[name] = data
        valid_mask = valid if valid_mask is None else valid_mask & valid

# Convert aspect degrees into sine and cosine
aspect_radians = np.deg2rad(arrays["Aspect"])

arrays["Aspect_sin"] = np.sin(aspect_radians)
arrays["Aspect_cos"] = np.cos(aspect_radians)

# Arrange features in the exact training order
feature_names = [
    "DEM_Elevation",
    "DEM_Slope",
    "Aspect_sin",
    "Aspect_cos",
    "DEM_PlanCurv",
    "DEM_ProfCurv",
]

# Create prediction array
prediction = np.full(
    arrays["DEM_Elevation"].shape,
    np.nan,
    dtype=np.float32,
)

# Predict in batches to limit memory usage
rows, cols = np.where(valid_mask)

batch_size = 100_000

for start in range(0, len(rows), batch_size):
    end = start + batch_size

    batch = pd.DataFrame({
        feature: arrays[feature][rows[start:end], cols[start:end]]
        for feature in feature_names
    })

    prediction[rows[start:end], cols[start:end]] = (
        model.predict_proba(batch)[:, 1]
    )

# Save final raster
output_path = Path(
    "../data/processed/final_hgb_susceptibility.tif"
)

profile.update(
    driver="GTiff",
    dtype="float32",
    count=1,
    nodata=-9999.0,
    compress="lzw",
)

with rasterio.open(output_path, "w", **profile) as dst:
    dst.write(
        np.where(valid_mask, prediction, -9999.0).astype(np.float32),
        1,
    )

print("Final susceptibility raster saved:", output_path)
print("Valid prediction pixels:", len(rows))
print("Minimum score:", np.nanmin(prediction))
print("Maximum score:", np.nanmax(prediction))
print("Mean score:", np.nanmean(prediction))
```

```python
import geopandas as gpd
import rasterio
from rasterio.mask import mask
from pathlib import Path

# Input files
raster_path = "../data/processed/final_hgb_susceptibility.tif"
boundary_path = "../data/processed/study_area_boundary.geojson"

# Output file
output_path = Path(
    "../data/processed/final_hgb_susceptibility_clipped.tif"
)

# Load study-area boundary
boundary = gpd.read_file(boundary_path)

# Open the final susceptibility raster
with rasterio.open(raster_path) as src:

    # Reproject boundary to match raster CRS
    boundary = boundary.to_crs(src.crs)

    # Clip raster using the study-area polygon
    clipped_data, clipped_transform = mask(
        src,
        boundary.geometry,
        crop=True,
        nodata=-9999.0
    )

    # Update raster metadata
    clipped_profile = src.profile.copy()
    clipped_profile.update(
        height=clipped_data.shape[1],
        width=clipped_data.shape[2],
        transform=clipped_transform,
        nodata=-9999.0,
        compress="lzw"
    )

    # Save clipped raster
    with rasterio.open(output_path, "w", **clipped_profile) as dst:
        dst.write(clipped_data)

print("Clipped raster saved:", output_path)
print("Output dimensions:", clipped_data.shape[2], "x", clipped_data.shape[1])
print("Valid pixels:", (clipped_data[0] != -9999.0).sum())
```

## Spatial Coverage Assessment of Landslide Inventory

```python
import rasterio
import numpy as np
from rasterio.features import geometry_mask
from shapely.geometry import mapping

coverage_results = []

with rasterio.open(raster_path) as src:
    raster = src.read(1, masked=True)

    for idx, geometry in enumerate(gdf.geometry):
        # Restrict processing to the polygon's bounding window.
        try:
            window = rasterio.windows.from_bounds(
                *geometry.bounds,
                transform=src.transform
            )
            window = window.round_offsets().round_lengths()

            row_start = max(0, int(window.row_off))
            col_start = max(0, int(window.col_off))
            row_stop = min(src.height, row_start + int(window.height))
            col_stop = min(src.width, col_start + int(window.width))

            if row_start >= row_stop or col_start >= col_stop:
                default_count = 0
                touched_count = 0
            else:
                window_data = raster[
                    row_start:row_stop,
                    col_start:col_stop
                ]

                transform = src.window_transform(
                    rasterio.windows.Window(
                        col_start,
                        row_start,
                        col_stop - col_start,
                        row_stop - row_start
                    )
                )

                default_mask = geometry_mask(
                    [mapping(geometry)],
                    out_shape=window_data.shape,
                    transform=transform,
                    invert=True,
                    all_touched=False
                )

                touched_mask = geometry_mask(
                    [mapping(geometry)],
                    out_shape=window_data.shape,
                    transform=transform,
                    invert=True,
                    all_touched=True
                )

                valid = ~np.ma.getmaskarray(window_data)

                default_count = int(np.sum(default_mask & valid))
                touched_count = int(np.sum(touched_mask & valid))

            coverage_results.append({
                "sample_index": idx,
                "default_valid_pixels": default_count,
                "touched_valid_pixels": touched_count
            })

        except Exception as e:
            coverage_results.append({
                "sample_index": idx,
                "default_valid_pixels": 0,
                "touched_valid_pixels": 0
            })

coverage_df = pd.DataFrame(coverage_results)

print("FULL INVENTORY RASTER COVERAGE")
print("--------------------------------")
print("Total polygons:", len(coverage_df))
print(
    "Polygons with default valid pixels:",
    (coverage_df["default_valid_pixels"] > 0).sum()
)
print(
    "Polygons with only boundary-touching coverage:",
    (
        (coverage_df["default_valid_pixels"] == 0) &
        (coverage_df["touched_valid_pixels"] > 0)
    ).sum()
)
print(
    "Polygons with no valid raster coverage:",
    (coverage_df["touched_valid_pixels"] == 0).sum()
)
```
